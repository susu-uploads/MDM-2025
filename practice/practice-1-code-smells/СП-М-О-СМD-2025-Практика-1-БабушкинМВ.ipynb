2. Реализуйте любой алгоритм сортировки на любом языке программирования.

```python
# Сортировка выбором

def selection_sort(arr):
    length = len(arr)

    for i in range(length):
        min_idx = i
        for j in range(i + 1, length):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]

    return arr

arr = [4, 2, 3, 7, 8, 4, 9]
print(selection_sort(arr))
```

```python
# Исправленная версия

def bubble_sort(arr):
    length = len(arr)
    for i in range(length):
        changed = False
        for j in range(0, length - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                changed = True
        if not changed:
            break
    return arr
```

```python
# Изначальная правильная версия Чашина Дениса

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr
```
